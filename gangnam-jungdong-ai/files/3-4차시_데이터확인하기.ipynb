## 1. 파일이 잘 올라갔는지 확인하기
아래 셀을 실행해서 두 파일이 모두 ✅ 로 나오는지 확인합니다.

```python
import glob

정류소_파일 = sorted(glob.glob("*정류소*.xlsx"))
승하차_파일 = sorted(glob.glob("*승하차*.csv"))

print("✅ 정류소 위치정보 파일:", 정류소_파일[0]) if 정류소_파일 else print("❌ 정류소 위치정보(.xlsx) 파일이 없어요. 0번 단계를 다시 해 주세요.")
print("✅ 승하차 인원 파일:", 승하차_파일[0]) if 승하차_파일 else print("❌ 승하차 인원(.csv) 파일이 없어요. 0번 단계를 다시 해 주세요.")
```

## 2. 데이터 불러오기

`pandas`는 표 형태의 데이터를 다루는 파이썬 도구입니다.
- 정류장 데이터(엑셀 파일) → `stations` 라는 이름으로 저장
- 승하차 데이터(CSV 파일) → `data` 라는 이름으로 저장

> 승하차 CSV는 한글이 들어 있어 `encoding="cp949"` 로 읽습니다.

```python
import pandas as pd

stations = pd.read_excel(정류소_파일[0])

try:
    data = pd.read_csv(승하차_파일[0], encoding="cp949", low_memory=False)
except UnicodeDecodeError:
    data = pd.read_csv(승하차_파일[0], encoding="utf-8", low_memory=False)

print("🎉 데이터 불러오기 완료!")
print(f"정류장 데이터: {len(stations):,}줄  /  승하차 데이터: {len(data):,}줄")
```

## 5. (도전) 데이터 구조 더 살펴보기 🔍

시간이 남으면 아래 셀들도 실행해 봅니다. 5/6차시 데이터 분석의 준비 운동이에요.

```python
# 5-1. 데이터 크기: (행 수, 열 수)
print("정류장 데이터 크기:", stations.shape)
print("승하차 데이터 크기:", data.shape)
```

```python
# 5-2. 승하차 데이터의 속성(열) 이름 전체 보기
for i, col in enumerate(data.columns, start=1):
    print(i, col)
```

```python
# 5-3. 우리 동네 정류장 찾아보기 — 따옴표 안의 글자를 바꿔서 실행해 보세요!
찾을_이름 = "역삼"

이름_열 = [c for c in data.columns if "역명" in c or "정류장명" in c or "정류소명" in c]
if 이름_열:
    결과 = data[data[이름_열[0]].astype(str).str.contains(찾을_이름)]
    print(f"'{찾을_이름}'이(가) 들어간 정류장 데이터: {len(결과):,}줄")
    display(결과.head(10))
else:
    print("정류장 이름 열을 찾지 못했어요. 5-2의 속성 이름을 확인해 보세요.")
```
